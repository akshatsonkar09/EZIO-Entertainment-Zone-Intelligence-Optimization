### RRF

```python
def reciprocal_rank_fusion(result_lists, k=60):
    scores = {}

    for result_list in result_lists:
        for rank, item in enumerate(result_list, start=1):
            scores[item] = scores.get(item, 0) + (
                1 / (k + rank)
            )

    ranked_results = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return ranked_results
```

### Cross-Encoder Reranking

```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder(
    "cross-encoder/ms-marco-MiniLM-L-6-v2"
)

print("Cross-Encoder loaded.")
```

```python
def rerank_candidates(query, candidates, top_k=10):

    pairs = [
        (query, candidate["document"])
        for candidate in candidates
    ]

    scores = reranker.predict(pairs)

    reranked = []

    for candidate, score in zip(candidates, scores):
        reranked.append({
            "id": candidate["id"],
            "document": candidate["document"],
            "score": float(score),
            "media_type": candidate["media_type"]
        })

    reranked.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return reranked[:top_k]
```

```python
def get_valid_ids(dataframe, current_ids, filters, media_type):

    valid_indices = []

    for idx, row in dataframe.iterrows():

        # -------------------------
        # Common: Release Year
        # -------------------------

        if filters.get("release_year_min") is not None:
            if (
                pd.isna(row["release_year"])
                or row["release_year"] < filters["release_year_min"]
            ):
                continue

        if filters.get("release_year_max") is not None:
            if (
                pd.isna(row["release_year"])
                or row["release_year"] > filters["release_year_max"]
            ):
                continue


        # -------------------------
        # Common: Genres
        # -------------------------

        genre_column = {
            "movie": "genres",
            "anime": "genres",
            "music": "playlist_genre",
            "book": "genres"
        }[media_type]

        row_genres = str(row.get(genre_column, "")).lower()


        if filters.get("genres_include"):
            if not all(
                genre.lower() in row_genres
                for genre in filters["genres_include"]
            ):
                continue


        if filters.get("genres_exclude"):
            if any(
                genre.lower() in row_genres
                for genre in filters["genres_exclude"]
            ):
                continue


        # -------------------------
        # Movie-specific filters
        # -------------------------

        if media_type == "movie":

            if filters.get("runtime_min") is not None:
                if (
                    pd.isna(row["runtime"])
                    or row["runtime"] < filters["runtime_min"]
                ):
                    continue

            if filters.get("runtime_max") is not None:
                if (
                    pd.isna(row["runtime"])
                    or row["runtime"] > filters["runtime_max"]
                ):
                    continue

            if filters.get("revenue_min") is not None:
                if (
                    pd.isna(row["revenue"])
                    or row["revenue"] < filters["revenue_min"]
                ):
                    continue


        # -------------------------
        # Anime-specific filters
        # -------------------------

        if media_type == "anime":

            if filters.get("episodes_min") is not None:
                if (
                    pd.isna(row["episodes"])
                    or row["episodes"] < filters["episodes_min"]
                ):
                    continue

            if filters.get("episodes_max") is not None:
                if (
                    pd.isna(row["episodes"])
                    or row["episodes"] > filters["episodes_max"]
                ):
                    continue

            if filters.get("popularity_min") is not None:
                if (
                    pd.isna(row["popularity"])
                    or row["popularity"] < filters["popularity_min"]
                ):
                    continue

            if filters.get("ranking_max") is not None:
                if (
                    pd.isna(row["ranking"])
                    or row["ranking"] > filters["ranking_max"]
                ):
                    continue


        # -------------------------
        # Music-specific filters
        # -------------------------

        if media_type == "music":

            if filters.get("track_name") is not None:
                if str(row["track_name"]).lower() != filters["track_name"].lower():
                    continue

            if filters.get("artist") is not None:
                if str(row["artist"]).lower() != filters["artist"].lower():
                    continue

            if filters.get("album") is not None:
                if str(row["album"]).lower() != filters["album"].lower():
                    continue

            if filters.get("playlist_genre") is not None:
                if (
                    str(row["playlist_genre"]).lower()
                    != filters["playlist_genre"].lower()
                ):
                    continue

            if filters.get("playlist_subgenre") is not None:
                if (
                    str(row["playlist_subgenre"]).lower()
                    != filters["playlist_subgenre"].lower()
                ):
                    continue

            if filters.get("duration_min") is not None:
                if (
                    pd.isna(row["duration_minutes"])
                    or row["duration_minutes"] < filters["duration_min"]
                ):
                    continue

            if filters.get("duration_max") is not None:
                if (
                    pd.isna(row["duration_minutes"])
                    or row["duration_minutes"] > filters["duration_max"]
                ):
                    continue


        # -------------------------
        # Book-specific filters
        # -------------------------

        if media_type == "book":

            if filters.get("rating_min") is not None:
                if (
                    pd.isna(row["rating"])
                    or row["rating"] < filters["rating_min"]
                ):
                    continue

            if filters.get("pages_min") is not None:
                if (
                    pd.isna(row["pages"])
                    or row["pages"] < filters["pages_min"]
                ):
                    continue

            if filters.get("pages_max") is not None:
                if (
                    pd.isna(row["pages"])
                    or row["pages"] > filters["pages_max"]
                ):
                    continue

            if filters.get("author"):
                if not any(
                    author.lower() in str(row["author"]).lower()
                    for author in filters["author"]
                ):
                    continue

            if filters.get("publisher") is not None:
                if (
                    filters["publisher"].lower()
                    not in str(row["publisher"]).lower()
                ):
                    continue


        valid_indices.append(idx)


    return {
        current_ids[i]
        for i in valid_indices
    }
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
def hybrid_retrieve(
    query,
    media_type,
    filters=None,
    top_k=10,
    candidate_k=30
):

    if filters is None:
        filters = {}

    # --------------------------------
    # 1. GET RESOURCES
    # --------------------------------

    collections = {
        "movie": movie_collection,
        "anime": anime_collection,
        "music": music_collection,
        "book": book_collection
    }

    bm25_indexes = {
        "movie": movie_bm25,
        "anime": anime_bm25,
        "music": music_bm25,
        "book": book_bm25
    }

    ids = {
        "movie": movie_ids,
        "anime": anime_ids,
        "music": music_ids,
        "book": book_ids
    }

    documents = {
        "movie": movie_documents,
        "anime": anime_documents,
        "music": music_documents,
        "book": book_documents
    }

    dataframes = {
        "movie": movie_metadata_clean,
        "anime": anime_metadata_clean,
        "music": music_metadata_clean,
        "book": book_metadata_clean
    }

    title_fields = {
        "movie": "title",
        "anime": "title",
        "music": "track_name",
        "book": "title"
    }

    # --------------------------------
    # 2. VALIDATE MEDIA TYPE
    # --------------------------------

    if media_type not in collections:
        raise ValueError(
            f"Unsupported media type: {media_type}"
        )

    collection = collections[media_type]
    bm25_index = bm25_indexes[media_type]
    current_ids = ids[media_type]
    current_documents = documents[media_type]
    dataframe = dataframes[media_type]
    title_field = title_fields[media_type]

    # Map ID → dataframe index
    id_to_index = {
        item_id: idx
        for idx, item_id in enumerate(current_ids)
    }

    # --------------------------------
    # 3. METADATA FILTERING
    # --------------------------------

    valid_ids = get_valid_ids(
        dataframe=dataframe,
        current_ids=current_ids,
        filters=filters,
        media_type=media_type
    )

    if not valid_ids:
        return []

    # --------------------------------
    # 4. SEMANTIC RETRIEVAL
    # --------------------------------

    query_embedding = model.encode([query])

    semantic_candidate_k = min(
        max(candidate_k * 5, 100),
        len(current_ids)
    )

    semantic_result = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=semantic_candidate_k
    )

    semantic_ids_raw = semantic_result["ids"][0]

    # Apply metadata filtering
    semantic_ids = [
        item_id
        for item_id in semantic_ids_raw
        if item_id in valid_ids
    ]

    semantic_ids = semantic_ids[:candidate_k]

    # --------------------------------
    # 5. BM25 RETRIEVAL
    # --------------------------------

    query_tokens = tokenize(query)

    bm25_scores = bm25_index.get_scores(
        query_tokens
    )

    bm25_indices = bm25_scores.argsort()[::-1]

    bm25_ids = []

    for idx in bm25_indices:

        item_id = current_ids[idx]

        if item_id in valid_ids:
            bm25_ids.append(item_id)

        if len(bm25_ids) >= candidate_k:
            break

    # --------------------------------
    # 6. RECIPROCAL RANK FUSION
    # --------------------------------

    rrf_results = reciprocal_rank_fusion(
        [
            semantic_ids,
            bm25_ids
        ]
    )

    candidate_ids = [
        item_id
        for item_id, score in rrf_results[:candidate_k]
    ]

    if not candidate_ids:
        return []

    # --------------------------------
    # 7. BUILD CANDIDATE DOCUMENTS
    # --------------------------------

    candidate_documents = [
        current_documents[id_to_index[item_id]]
        for item_id in candidate_ids
    ]

    # --------------------------------
    # 8. CROSS-ENCODER RERANKING
    # --------------------------------

    pairs = [
        [query, document]
        for document in candidate_documents
    ]

    rerank_scores = reranker.predict(pairs)

    reranked = sorted(
        zip(candidate_ids, rerank_scores),
        key=lambda x: x[1],
        reverse=True
    )

    # --------------------------------
    # 9. FINAL RESULTS
    # --------------------------------

    results = []

    for item_id, score in reranked[:top_k]:

        idx = id_to_index[item_id]

        results.append({
            "id": item_id,
            "title": dataframe.iloc[idx][title_field],
            "score": float(score),
            "media_type": media_type
        })

    return results
```

```python
import sys
from pathlib import Path

sys.path.append(str(Path("../src").resolve()))

from project_black_knight.query_parser import parse_query


def search_black_knight(user_query: str, top_k: int = 10):

    """
    End-to-end Project Black Knight search.

    User query
        -> LLM query understanding
        -> structured filters
        -> hybrid retrieval
        -> cross-encoder reranking
    """

    # --------------------------------
    # 1. LLM QUERY UNDERSTANDING
    # --------------------------------

    parsed_query = parse_query(user_query)


    # --------------------------------
    # 2. PRIMARY MEDIA TYPE
    # --------------------------------

    if not parsed_query.media_types:
        raise ValueError(
            "No media type detected in query."
        )

    primary_media = parsed_query.media_types[0]


    # --------------------------------
    # 3. COMMON CONSTRAINTS
    # --------------------------------

    common = parsed_query.common


    # --------------------------------
    # 4. DOMAIN-SPECIFIC CONSTRAINTS
    # --------------------------------

    domain_queries = parsed_query.domain_queries

    if primary_media == "movie":
        domain_query = domain_queries.movie

    elif primary_media == "anime":
        domain_query = domain_queries.anime

    elif primary_media == "music":
        domain_query = domain_queries.music

    elif primary_media == "book":
        domain_query = domain_queries.book

    else:
        raise ValueError(
            f"Unsupported media type: {primary_media}"
        )


    # --------------------------------
    # 5. BUILD FILTERS
    # --------------------------------

    filters = {
        "genres_include": common.genres_include,
        "genres_exclude": common.genres_exclude,
        "release_year_min": common.release_year_min,
        "release_year_max": common.release_year_max,
    }


    # --------------------------------
    # 6. ADD DOMAIN-SPECIFIC FILTERS
    # --------------------------------

    if domain_query is not None:

        domain_filters = domain_query.model_dump(
            exclude_none=True
        )

        filters.update(domain_filters)


    # --------------------------------
    # 7. SEMANTIC QUERY
    # --------------------------------

    semantic_query = parsed_query.semantic_query


    # --------------------------------
    # 8. HYBRID RETRIEVAL
    # --------------------------------

    results = hybrid_retrieve(
        query=semantic_query,
        media_type=primary_media,
        filters=filters,
        top_k=top_k
    )


    # --------------------------------
    # 9. RETURN
    # --------------------------------

    return {
        "parsed_query": parsed_query,
        "results": results
    }
```

```python
query = """
Recommend me mind-bending science fiction movies and anime,
released after 2010, with no horror.
I want movies under 150 minutes and anime with no more than 50 episodes.
Something similar in feel to Inception or Steins;Gate.
"""

result = search_black_knight(query, top_k=10)

print("\n========== PARSED QUERY ==========")
print(result["parsed_query"])

print("\n========== RESULTS ==========")

for i, item in enumerate(result["results"], start=1):
    print(
        f"{i}. {item['title']} "
        f"| {item['media_type']} "
        f"| score={item['score']:.4f}"
    )
```
