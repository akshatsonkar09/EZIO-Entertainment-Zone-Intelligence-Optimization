```python
from difflib import SequenceMatcher
from sklearn.metrics.pairwise import cosine_similarity

def recommend_semantic_integrated(movie, n=5):

    #Normalizing the inputs AvaTar to avatar so all combination of capitalization works
    movienorm = normalizeinput(movie)

    # If movie name is not found that is typo or something like avtr and stuff then this fuzzy logic is going to work
    if movienorm not in movies['title_normalized'].values:

        best_match = None
        best_score = 0

        for title in movies['title_normalized']:
            score = SequenceMatcher(None, movienorm, title).ratio()

            if score > best_score:
                best_score = score
                best_match = title

        # THIS CUTOFF IS GOING TO DECIDE THE SWITCH BETWEEN SEMANTIC SEARCH AND CONTENT BASED FILTERING
        if best_score >= 0.85:

            #If close match is found then use that close match
            movienorm = best_match

            # Finding movie title
            matched_title = movies.loc[
                movies['title_normalized'] == movienorm,
                'title'
            ].iloc[0]

            print(f"Using closest match: {matched_title}")

        else:
            # If still no close match then use semantic search
            
            # Encoding users new input
            query_embedding = model.encode([movie])

            semantic_scores = cosine_similarity(
                query_embedding,
                movie_contextual_embeddings
            )[0]

            # Taking top 5 similarity scores
            top_indices = semantic_scores.argsort()[::-1][:n]

            result = []

            for i in top_indices:
                result.append({
                    'title': movies.iloc[i].title,
                    'similarity_score': semantic_scores[i]
                })

            return pd.DataFrame(result)


    # Finding movie index
    movie_index = movies[
        movies['title_normalized'] == movienorm
    ].index[0]

    # Checking the similarity scores
    distance = contextual_similarity_matrix[movie_index]

    # Taking top 5 similarity scores
    movies_list = sorted(
        list(enumerate(distance)),
        reverse=True,
        key=lambda x: x[1]
    )[1:n+1]


    result = []

    for i in movies_list:
        result.append({
            'title': movies.iloc[i[0]].title,
            'similarity_score': i[1]
        })

    return pd.DataFrame(result)
```
