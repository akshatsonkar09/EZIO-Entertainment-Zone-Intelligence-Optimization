```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
songs_contextual_embeddings = model.encode(songs["tag"].tolist(),)
books_contextual_embeddings = model.encode(books["tag2"].tolist(),)
anime_contextual_embeddings = model.encode(anime["tag2"].tolist(),)
```

```python
print("Song" ,songs_contextual_embeddings.shape)
print("Books" ,books_contextual_embeddings.shape)
print("Anime" ,anime_contextual_embeddings.shape)

print(songs_contextual_embeddings[0].shape)
print(books_contextual_embeddings[0].shape)
print(anime_contextual_embeddings[0].shape)
```

```python
from pathlib import Path
embedding_dir = Path(r"C:\Coding\Project Black Knight\data\processed\embeddings")
embedding_dir.mkdir(parents=True, exist_ok=True)
```

```python
np.save(embedding_dir / "songs_embeddings.npy",songs_contextual_embeddings)
np.save(embedding_dir / "books_embeddings.npy",books_contextual_embeddings)
np.save(embedding_dir / "anime_embeddings.npy",anime_contextual_embeddings)
```

```python
movie_contextual_embeddings = model.encode(
    movies["tag"].tolist(),
    show_progress_bar=True
)

print(movie_contextual_embeddings.shape)
```

```python
np.save(embedding_dir / "movie_embeddings.npy",movie_contextual_embeddings)
```
