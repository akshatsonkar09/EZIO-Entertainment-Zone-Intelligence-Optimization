```python
import pandas as pd
import numpy as np
import ast


book_metadata = pd.read_csv(
    r"..\data\processed\book_metadata.csv"
)


def clean_text(value):
    if value is None or pd.isna(value):
        return ""

    # Handle list-like values such as "['Fantasy', 'Adventure']"
    if isinstance(value, str):
        value = value.strip()

        if value.startswith("[") and value.endswith("]"):
            try:
                parsed = ast.literal_eval(value)

                if isinstance(parsed, list):
                    return ", ".join(
                        str(x) for x in parsed
                        if x is not None
                    )

            except (ValueError, SyntaxError):
                pass

        return value

    return str(value)


def clean_number(value):
    if value is None or pd.isna(value):
        return None

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    return value


# Text fields
book_metadata["Title"] = book_metadata["Title"].apply(clean_text)
book_metadata["Author"] = book_metadata["Author"].apply(clean_text)
book_metadata["genres_clean"] = book_metadata["genres_clean"].apply(clean_text)
book_metadata["Characters"] = book_metadata["Characters"].apply(clean_text)
book_metadata["Publisher"] = book_metadata["Publisher"].apply(clean_text)
book_metadata["PublishDate"] = book_metadata["PublishDate"].apply(clean_text)

# Numeric fields
book_metadata["Rating"] = book_metadata["Rating"].apply(clean_number)
book_metadata["Pages"] = book_metadata["Pages"].apply(clean_number)



def clean_isbn(value):
    if value is None or pd.isna(value):
        return ""

    value = str(value)

    if value.endswith(".0"):
        value = value[:-2]

    return value


book_metadata["Isbn"] = book_metadata["Isbn"].apply(clean_isbn)



book_metadata_records = book_metadata.to_dict(
    orient="records"
)


for record in book_metadata_records:
    for key in list(record.keys()):
        if record[key] is None:
            del record[key]

# --- Standardize columns + add release_year (needed by validation and
# by the common release_year filter that applies across all media types) ---

book_metadata = book_metadata.rename(columns={
    "Title": "title",
    "Author": "author",
    "Rating": "rating",
    "Isbn": "isbn",
    "genres_clean": "genres",
    "Characters": "characters",
    "Pages": "pages",
    "Publisher": "publisher",
    "PublishDate": "publish_date",
})

def extract_year(value):
    if value is None or pd.isna(value):
        return None
    match = re.search(r"(\d{4})", str(value))
    return int(match.group(1)) if match else None

import re
book_metadata["release_year"] = book_metadata["publish_date"].apply(extract_year)

book_metadata_records = book_metadata.to_dict(orient="records")

for record in book_metadata_records:
    for key in list(record.keys()):
        if record[key] is None:
            del record[key]

book_metadata.to_csv(r"..\data\processed\book_metadata_clean.csv", index=False)
```

```python
movie_metadata = pd.read_csv(
    r"..\data\processed\movie_metadata.csv"
)

import ast
import pandas as pd
import numpy as np


def clean_movie_list(value):

    if value is None or pd.isna(value):
        return ""

    if isinstance(value, str):

        value = value.strip()

        if value.startswith("[") and value.endswith("]"):
            try:
                parsed = ast.literal_eval(value)

                if isinstance(parsed, list):

                    result = []

                    for item in parsed:

                        # Simple strings
                        if isinstance(item, str):
                            result.append(item)

                        # Dictionaries such as production companies
                        elif isinstance(item, dict):
                            if "name" in item:
                                result.append(str(item["name"]))

                    return ", ".join(result)

            except (ValueError, SyntaxError):
                pass

        return value

    return str(value)


def clean_text(value):

    if value is None or pd.isna(value):
        return ""

    return str(value)


def clean_number(value):

    if value is None or pd.isna(value):
        return None

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    return value

movie_metadata["genres"] = movie_metadata["genres"].apply(
    clean_movie_list
)

movie_metadata["production_companies"] = movie_metadata[
    "production_companies"
].apply(clean_movie_list)

movie_metadata["cast"] = movie_metadata["cast"].apply(
    clean_movie_list
)

movie_metadata["title"] = movie_metadata["title"].apply(
    clean_text
)

movie_metadata["release_date"] = movie_metadata[
    "release_date"
].apply(clean_text)

movie_metadata["director"] = movie_metadata["director"].apply(
    clean_movie_list
)

movie_metadata["revenue"] = movie_metadata["revenue"].apply(
    clean_number
)

movie_metadata["runtime"] = movie_metadata["runtime"].apply(
    clean_number
)


movie_metadata["release_year"] = pd.to_datetime(
    movie_metadata["release_date"],
    errors="coerce"
).dt.year


movie_metadata["release_year"] = movie_metadata[
    "release_year"
].apply(clean_number)



movie_metadata_records = movie_metadata.to_dict(
    orient="records"
)


for record in movie_metadata_records:
    for key in list(record.keys()):
        if record[key] is None:
            del record[key]





movie_metadata.to_csv(r"..\data\processed\movie_metadata_clean.csv", index=False)
```

```python
allowed_types = (str, int, float, bool)

def validate_records(records):

    invalid = []

    for i, record in enumerate(records):

        for key, value in record.items():

            if not isinstance(value, allowed_types):
                invalid.append(
                    (i, key, type(value).__name__, value)
                )

    return invalid


book_invalid = validate_records(book_metadata_records)
movie_invalid = validate_records(movie_metadata_records)

print("Invalid book values:", len(book_invalid))
print("Invalid movie values:", len(movie_invalid))
```

```python
def reciprocal_rank_fusion(result_lists, k=60):
    scores = {}

    for result_list in result_lists:
        for rank, item in enumerate(result_list, start=1):
            scores[item] = scores.get(item, 0) + (
                1 / (k + rank)
            )

    ranked_results = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    return ranked_results
```

```python
def hybrid_retrieve(query, top_k=10, candidate_k=30):

    # -------------------------
    # 1. Semantic retrieval
    # -------------------------

    query_embedding = model.encode([query])

    semantic_result = movie_collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=candidate_k
    )

    semantic_ids = semantic_result["ids"][0]


    # -------------------------
    # 2. BM25 retrieval
    # -------------------------

    query_tokens = tokenize(query)

    bm25_scores = movie_bm25.get_scores(query_tokens)

    bm25_indices = bm25_scores.argsort()[::-1][:candidate_k]

    bm25_ids = [
        movie_ids[i]
        for i in bm25_indices
    ]


    # -------------------------
    # 3. RRF
    # -------------------------

    rrf_results = reciprocal_rank_fusion(
        [semantic_ids, bm25_ids]
    )


    # -------------------------
    # 4. Cross-encoder
    # -------------------------

    candidate_ids = [
        movie_id
        for movie_id, score in rrf_results[:candidate_k]
    ]

    candidate_indices = [
        movie_ids.index(movie_id)
        for movie_id in candidate_ids
    ]

    candidate_documents = [
        movie_documents[i]
        for i in candidate_indices
    ]

    pairs = [
        [query, document]
        for document in candidate_documents
    ]

    rerank_scores = reranker.predict(pairs)

    reranked = sorted(
        zip(candidate_ids, rerank_scores),
        key=lambda x: x[1],
        reverse=True
    )


    # -------------------------
    # 5. Return final results
    # -------------------------

    results = []

    for movie_id, score in reranked[:top_k]:

        idx = movie_ids.index(movie_id)

        results.append({
            "movie_id": movie_id,
            "title": movies.iloc[idx]["title"],
            "score": float(score)
        })

    return results
```

```python
def build_chroma_filter(filters, media_type):
    conditions = []

    if filters.get("release_year_min") is not None:
        conditions.append({
            "release_year": {
                "$gte": filters["release_year_min"]
            }
        })

    if filters.get("release_year_max") is not None:
        conditions.append({
            "release_year": {
                "$lte": filters["release_year_max"]
            }
        })

    # Domain-specific filters will be added here
    # after the common filters are working.

    if not conditions:
        return None

    if len(conditions) == 1:
        return conditions[0]

    return {"$and": conditions}
```

```python
def row_matches_filters(row, filters, media_type):
    """
    Check whether a dataframe row satisfies all requested filters.
    """

    # -------------------------
    # COMMON FILTERS
    # -------------------------

    release_year_min = filters.get("release_year_min")
    if release_year_min is not None:
        if row["release_year"] < release_year_min:
            return False

    release_year_max = filters.get("release_year_max")
    if release_year_max is not None:
        if row["release_year"] > release_year_max:
            return False

    # -------------------------
    # GENRE FILTERS
    # -------------------------

    genres_include = filters.get("genres_include", [])
    genres_exclude = filters.get("genres_exclude", [])

    # Find the appropriate genre column
    genre_column = {
        "movie": "genres",
        "anime": "genres",
        "music": "playlist_genre",
        "book": "genres"
    }[media_type]

    row_genres = str(row[genre_column]).lower()

    # Every requested genre must be present
    for genre in genres_include:
        if genre.lower() not in row_genres:
            return False

    # None of the excluded genres should be present
    for genre in genres_exclude:
        if genre.lower() in row_genres:
            return False

    # -------------------------
    # DOMAIN-SPECIFIC FILTERS
    # -------------------------

    domain_query = filters.get("domain")

    if domain_query is None:
        return True

    # Convert Pydantic model → dictionary
    if hasattr(domain_query, "model_dump"):
        domain_query = domain_query.model_dump(exclude_none=True)

    # =========================
    # MOVIE
    # =========================

    if media_type == "movie":

        production_companies = domain_query.get(
            "production_companies", []
        )

        for company in production_companies:
            value = str(row["production_companies"]).lower()

            if company.lower() not in value:
                return False

        cast = domain_query.get("cast", [])

        for actor in cast:
            value = str(row["cast"]).lower()

            if actor.lower() not in value:
                return False

        directors = domain_query.get("director", [])

        for director in directors:
            value = str(row["director"]).lower()

            if director.lower() not in value:
                return False

        revenue_min = domain_query.get("revenue_min")

        if revenue_min is not None:
            if row["revenue"] < revenue_min:
                return False

        runtime_min = domain_query.get("runtime_min")

        if runtime_min is not None:
            if row["runtime"] < runtime_min:
                return False

        runtime_max = domain_query.get("runtime_max")

        if runtime_max is not None:
            if row["runtime"] > runtime_max:
                return False

    # =========================
    # ANIME
    # =========================

    elif media_type == "anime":

        episodes_min = domain_query.get("episodes_min")

        if episodes_min is not None:
            if row["episodes"] < episodes_min:
                return False

        episodes_max = domain_query.get("episodes_max")

        if episodes_max is not None:
            if row["episodes"] > episodes_max:
                return False

        popularity_min = domain_query.get("popularity_min")

        if popularity_min is not None:
            if row["popularity"] < popularity_min:
                return False

        ranking_max = domain_query.get("ranking_max")

        if ranking_max is not None:
            if row["ranking"] > ranking_max:
                return False

    # =========================
    # MUSIC
    # =========================

    elif media_type == "music":

        track_id = domain_query.get("track_id")

        if track_id is not None:
            if str(row["track_id"]).lower() != track_id.lower():
                return False

        track_name = domain_query.get("track_name")

        if track_name is not None:
            if track_name.lower() not in str(row["track_name"]).lower():
                return False

        artist = domain_query.get("artist")

        if artist is not None:
            if artist.lower() not in str(row["artist"]).lower():
                return False

        album = domain_query.get("album")

        if album is not None:
            if album.lower() not in str(row["album"]).lower():
                return False

        playlist_genre = domain_query.get("playlist_genre")

        if playlist_genre is not None:
            if playlist_genre.lower() != str(
                row["playlist_genre"]
            ).lower():
                return False

        playlist_subgenre = domain_query.get("playlist_subgenre")

        if playlist_subgenre is not None:
            if playlist_subgenre.lower() != str(
                row["playlist_subgenre"]
            ).lower():
                return False

        duration_min = domain_query.get("duration_min")

        if duration_min is not None:
            if row["duration_minutes"] < duration_min:
                return False

        duration_max = domain_query.get("duration_max")

        if duration_max is not None:
            if row["duration_minutes"] > duration_max:
                return False

    # =========================
    # BOOK
    # =========================

    elif media_type == "book":

        authors = domain_query.get("author", [])

        row_author = str(row["author"]).lower()

        for author in authors:
            if author.lower() not in row_author:
                return False

        rating_min = domain_query.get("rating_min")

        if rating_min is not None:
            if row["rating"] < rating_min:
                return False

        isbn = domain_query.get("isbn")

        if isbn is not None:
            if str(row["isbn"]).lower() != isbn.lower():
                return False

        characters = domain_query.get("characters", [])

        row_characters = str(row["characters"]).lower()

        for character in characters:
            if character.lower() not in row_characters:
                return False

        pages_min = domain_query.get("pages_min")

        if pages_min is not None:
            if row["pages"] < pages_min:
                return False

        pages_max = domain_query.get("pages_max")

        if pages_max is not None:
            if row["pages"] > pages_max:
                return False

        publisher = domain_query.get("publisher")

        if publisher is not None:
            if publisher.lower() not in str(
                row["publisher"]
            ).lower():
                return False

    return True
```

```python
def get_valid_indices(dataframe, filters, media_type):
    valid_indices = []

    for idx, row in dataframe.iterrows():

        if row_matches_filters(row, filters, media_type ):
            valid_indices.append(idx)

    return valid_indices
```

```python
def get_valid_ids(dataframe, current_ids, filters, media_type):

    valid_indices = get_valid_indices(dataframe, filters, media_type)

    valid_ids = {
        current_ids[idx]
        for idx in valid_indices
    }

    return valid_ids
```

```python
def hybrid_retrieve(
    query,
    media_type,
    filters=None,
    top_k=10,
    candidate_k=30
):

    if filters is None:
        filters = {}

    # --------------------------------
    # 1. GET RESOURCES
    # --------------------------------

    collections = {
        "movie": movie_collection,
        "anime": anime_collection,
        "music": music_collection,
        "book": book_collection
    }

    bm25_indexes = {
        "movie": movie_bm25,
        "anime": anime_bm25,
        "music": music_bm25,
        "book": book_bm25
    }

    ids = {
        "movie": movie_ids,
        "anime": anime_ids,
        "music": music_ids,
        "book": book_ids
    }

    documents = {
        "movie": movie_documents,
        "anime": anime_documents,
        "music": music_documents,
        "book": book_documents
    }

    dataframes = {
        "movie": movie_metadata,
        "anime": anime_metadata,
        "music": music_metadata,
        "book": book_metadata
    }

    # The "title" field is called something different per dataset
    title_fields = {
        "movie": "title",
        "anime": "title",
        "music": "track_name",
        "book": "title"
    }

    if media_type not in collections:
        raise ValueError(
            f"Unsupported media type: {media_type}"
        )

    collection = collections[media_type]
    bm25_index = bm25_indexes[media_type]
    current_ids = ids[media_type]
    current_documents = documents[media_type]
    dataframe = dataframes[media_type]

    # --------------------------------
    # 2. METADATA FILTERING
    # --------------------------------

    valid_ids = get_valid_ids(
        dataframe=dataframe,
        current_ids=current_ids,
        filters=filters,
        media_type=media_type
    )

    if not valid_ids:
        return []

    # --------------------------------
    # 3. SEMANTIC RETRIEVAL
    # --------------------------------

    query_embedding = model.encode([query])

    # Retrieve more candidates because
    # metadata filtering happens after
    # semantic retrieval.
    semantic_candidate_k = min(
        max(candidate_k * 5, 100),
        len(current_ids)
    )

    semantic_result = collection.query(
        query_embeddings=query_embedding.tolist(),
        n_results=semantic_candidate_k
    )

    semantic_ids_raw = semantic_result["ids"][0]

    # Keep only IDs satisfying metadata filters
    semantic_ids = [
        item_id
        for item_id in semantic_ids_raw
        if item_id in valid_ids
    ]

    # Keep only required number
    semantic_ids = semantic_ids[:candidate_k]

    # --------------------------------
    # 4. BM25 RETRIEVAL
    # --------------------------------

    query_tokens = tokenize(query)

    bm25_scores = bm25_index.get_scores(
        query_tokens
    )

    # Rank all BM25 documents
    bm25_indices = bm25_scores.argsort()[::-1]

    bm25_ids = []

    for idx in bm25_indices:

        item_id = current_ids[idx]

        # Metadata filtering happens BEFORE
        # BM25 results enter RRF.
        if item_id in valid_ids:
            bm25_ids.append(item_id)

        if len(bm25_ids) >= candidate_k:
            break

    # --------------------------------
    # 5. RRF
    # --------------------------------

    rrf_results = reciprocal_rank_fusion(
        [
            semantic_ids,
            bm25_ids
        ]
    )

    # --------------------------------
    # 6. CANDIDATES
    # --------------------------------

    candidate_ids = [
        item_id
        for item_id, score in rrf_results[:candidate_k]
    ]

    if not candidate_ids:
        return []

    candidate_indices = [
        current_ids.index(item_id)
        for item_id in candidate_ids
    ]

    candidate_documents = [
        current_documents[i]
        for i in candidate_indices
    ]

    # --------------------------------
    # 7. CROSS-ENCODER
    # --------------------------------

    pairs = [
        [query, document]
        for document in candidate_documents
    ]

    rerank_scores = reranker.predict(pairs)

    reranked = sorted(
        zip(candidate_ids, rerank_scores),
        key=lambda x: x[1],
        reverse=True
    )

    # --------------------------------
    # 8. FINAL RESULTS
    # --------------------------------

    results = []

    for item_id, score in reranked[:top_k]:

        idx = current_ids.index(item_id)

        results.append({
            "id": item_id,
            "title": dataframe.iloc[idx][title_fields[media_type]],
            "score": float(score),
            "media_type": media_type
        })

    return results
```

```python
import sys
from pathlib import Path


sys.path.append(str(Path("../src").resolve()))

# from project_black_knight import query_parser
from project_black_knight.query_parser import parse_query



def search_black_knight(user_query: str, top_k: int = 10):
    """
    End-to-end Project Black Knight search.

    User query
        -> LLM query understanding
        -> structured filters
        -> hybrid retrieval
        -> reranking
    """

    # 1. Understand the user's query using the LLM
    parsed_query = parse_query(user_query)

  
    # 2. Extract the primary media type
    if not parsed_query.media_types:
        raise ValueError("No media type detected in query.")

    primary_media = parsed_query.media_types[0]


    # 3. Extract common constraints
    common = parsed_query.common


    # 4. Extract domain-specific constraints

    domain_queries = parsed_query.domain_queries

    if primary_media == "movie":
        domain_query = domain_queries.movie

    elif primary_media == "anime":
        domain_query = domain_queries.anime

    elif primary_media == "music":
        domain_query = domain_queries.music

    elif primary_media == "book":
        domain_query = domain_queries.book

    else:
        raise ValueError(f"Unsupported media type: {primary_media}")

    # 5. Build filters for the existing retrieval system

    filters = {
        "genres_include": common.genres_include,
        "genres_exclude": common.genres_exclude,
        "release_year_min": common.release_year_min,
        "release_year_max": common.release_year_max,
    }

    # Add domain-specific filters
    if domain_query is not None:
        filters["domain"] = domain_query

    # 6. Semantic part of the query

    semantic_query = parsed_query.semantic_query

    # 7. Call your existing hybrid retrieval

    results = hybrid_retrieve(query=semantic_query, media_type=primary_media, filters=filters, top_k=top_k)

    return {
        "parsed_query": parsed_query,
        "results": results
    }
```
