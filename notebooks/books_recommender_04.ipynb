### New clean Tags making

```python
import pandas as pd

books = pd.read_pickle(r"..\data\processed\books.pkl")
```

```python
books.head(3)
```

```python
import ast
import re

def clean_text(value):
    if pd.isna(value):
        return ""

    value = str(value)

    # Remove HTML tags
    value = re.sub(r"<[^>]+>", " ", value)

    # Remove excessive whitespace and newlines
    value = re.sub(r"\s+", " ", value)

    return value.strip()


def clean_book_genres(value):
    if pd.isna(value):
        return ""

    try:
        genres = ast.literal_eval(value)

        if isinstance(genres, list):
            return ", ".join(str(g).strip() for g in genres)

    except (ValueError, SyntaxError):
        pass

    return str(value)


books["genres_clean"] = books["Genres"].apply(clean_book_genres)

books["description_clean"] = books["Description"].apply(clean_text)

books["tag2"] = (
    "Title: " + books["Title"].fillna("") +
    " Author: " + books["Author"].fillna("") +
    " Genres: " + books["genres_clean"] +
    " Description: " + books["description_clean"]
).str.lower()
```
