### Order of normalization matters

1) Expand (Fix slang/contractions) -> "didn't" should be "did not"
2) Clean (Remove web junk) -> like URLs, tags, Extra newlines and tabs
   -> We use Regular Expressions (Regex) for this. Think of regex as a "search and destroy" for text patterns.
3) Normalize Case (Lowercase everything)
4) Remove Noise (Punctuation/Special chars)
5) Tokenize (Split into words) -> creating of tags
6) Filter (Remove stopwords)
7) Stem / Lemmatize (Reduce to root form)
